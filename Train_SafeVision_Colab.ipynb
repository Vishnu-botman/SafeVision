This notebook trains **Model B (PPE Detector)** on the **CHV Dataset** using a free GPU in Google Colab.

- **Target Classes:** `person`, `vest`, `helmet`
- **Architecture:** YOLOv8 Nano / Small
- **Expected Training Time:** ~5–10 minutes on T4 GPU

```python
# 3. Upload or Mount Google Drive to access CHV_dataset or chv_yolo folder
# If you have chv_yolo.zip uploaded:
# !unzip -q chv_yolo.zip -d ./

import os
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 4. Launch Training with YOLOv8 on GPU
from ultralytics import YOLO

# Load pretrained YOLOv8n base model
model = YOLO('yolov8n.pt')

# Train for 50 epochs
results = model.train(
    data='chv_yolo/dataset.yaml',   # Update path if in Drive: '/content/drive/MyDrive/chv_yolo/dataset.yaml'
    epochs=50,
    batch=16,
    imgsz=640,
    device=0,
    project='safevision_colab',
    name='ppe_detector_run',
    save=True,
    val=True
)

print('Best weights saved at:', results.save_dir + '/weights/best.pt')
```

```python
# 5. Download the trained weights
from google.colab import files
files.download(results.save_dir + '/weights/best.pt')
```
